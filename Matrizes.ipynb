1. Lista: crie uma estrutura de repetição para fazer a leitura de 5 números inteiros e os armazene dentro de uma lista. Após a leitura, crie outra estrutura de repetição para somar todos os valores digitados.

```python
lista = []
for _ in range(1,6):
  valor = int(input('Digite um valor: '))
  lista.append(valor)
```

```python
lista
```

```python
soma = 0
for i in range(len(lista)):
  #print(lista[i])
  soma += lista[i]
print('Soma: ', soma)
```

```python
len(lista)      #tamanho da lista
```

```python
import numpy as np
np.array(lista).sum()
```

2. Dicionário: Crie um dicionário para armazenar o nemo e a nota de 3 alunos, fazendo a leitura dos valores por meio de uma estrutura de repetição. depois, crie uma nova estrutura de repetição para somar todas as notas e retonar a média.

```python
alunos = {}
for _ in range(1,4):
  nome = input('Digite o nome: ')
  nota = float(input('Digite a nota: '))
  alunos[nome] = nota
```

```python
alunos
```

```python
alunos['joao']
```

```python
alunos.values()
```

```python
soma = 0
for nota in alunos.values():
  #print(nota)
  soma += nota
print('Média: ', soma / 3)
```

matriz  np.array([[3,4,1],

```python
matriz = np.array([[3,4,1],
                   [3,1,5]])
matriz
```

```python
matriz.shape
```

```python
soma = 0
for i in range(matriz.shape[0]):
  for j in range(matriz.shape[1]):
    #print(matriz[i][j])
    soma += matriz[i][j]
  print('Soma: ', soma)
```
